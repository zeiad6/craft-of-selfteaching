## 1. عمليات النصوص الأساسية

النص في بايثون يدعم Unicode كاملاً، فالعربية تعمل مباشرة. والنصوص **غير قابلة للتعديل**؛ كل دالة تُرجع نصاً جديداً.

</div>

```python
s = "  Self-Teaching is a Craft  "
print(s.strip().lower())
print(s.split())
print("-".join(["a", "b", "c"]))
print("Craft" in s, s.find("Craft"), s.strip()[:4], s.strip()[::-1])
print("مرحباً بالعالم".upper(), len("مرحباً"))
print(f"{3.14159:.2f} | {1234567:,} | {'x':>5} | {0.256:.1%}")
```

## 2. التعابير النمطية (Regex)

التعبير النمطي لغة مصغّرة لوصف **أنماط النصوص**. استخدم نصوصاً خاماً `r"..."` كي لا تُفسَّر الشرطة المائلة.

| الرمز | المعنى |
| :-- | :-- |
| `\d` | رقم |
| `\w` | حرف أو رقم أو `_` |
| `+` / `*` / `?` | مرة فأكثر / صفر فأكثر / اختياري |
| `[...]` | أحد الأحرف |
| `(...)` | مجموعة التقاط |

</div>

```python
import re

text = "Contact: ali@example.com, sara.k@mail.org | Tel: 2026-11-09"

emails = re.findall(r"[\w.]+@[\w.]+\.\w+", text)
print(emails)

m = re.search(r"(\d{4})-(\d{2})-(\d{2})", text)
if m:
    year, month, day = m.groups()
    print(f"{year=} {month=} {day=}")

print(re.sub(r"\d", "#", "Phone 0501234567"))
```

### ⚠️ أخطاء شائعة

1. **محاولة تعديل نص:** `word[0] = "P"` تُسبب `TypeError`؛ النصوص غير قابلة للتعديل.
2. **الطمع (Greedy):** `.+` تلتهم أكبر قدر ممكن. استخدم `.+?` لأقصر تطابق.
3. **نسيان النص الخام:** اكتب `r"\d+"` وليس `"\d+"`.
4. **تعبير نمطي معقّد جداً:** إن تعذّرت قراءته فاستخدم `split` و`startswith` أو قسّمه بـ `re.VERBOSE`.

</div>

```python
import re

# النصوص غير قابلة للتعديل
word = "python"
try:
    word[0] = "P"
except TypeError as error:
    print("TypeError:", error)
print(word.capitalize())

# الطمع مقابل الكسل
html = "<b>bold</b> and <i>italic</i>"
print(re.findall(r"<.+>", html))
print(re.findall(r"<.+?>", html))

# النص الخام يحافظ على الشرطة المائلة
print(len("\n"), len(r"\n"))
```

### 📌 ملخص الفصل

- النصوص تدعم Unicode كاملاً وغير قابلة للتعديل.
- `f-strings` تنسّق الأرقام والمحاذاة والنسب.
- `re` تصف الأنماط: `\d` و`\w` و`+` و`?` والمجموعات.
- استخدم النص الخام `r"..."` مع التعابير النمطية.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| النص الخام (Raw string) | `r".."` لا يفسّر الشرطة المائلة |
| مجموعة التقاط (Capture group) | `(...)` تستخرج جزءاً من التطابق |
| الطمع (Greedy) | تطابق أطول ما يمكن |
| Unicode | معيار يشمل العربية والإيموجي |
