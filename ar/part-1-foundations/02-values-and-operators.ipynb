## 1. القيم وأنواعها (Values & Data Types)

**القيمة (Value)** هي أبسط وحدة بيانات: رقم، أو نص، أو صواب/خطأ. ولكل قيمة **نوع (Type)** يحدّد ما يمكنك فعله بها. لا تحفظ الأنواع، بل **اسأل بايثون نفسها** بالدالة `type()`: هذه هي روح التعلّم الذاتي.

</div>

```python
print(type(42))        # int   : عدد صحيح
print(type(3.14))      # float : عدد عشري
print(type("مرحباً"))   # str   : نص
print(type(True))      # bool  : قيمة منطقية
print(type(None))      # NoneType : لا قيمة
```

<div dir="rtl">

## 2. المتغيّرات (Variables)

**المتغيّر** اسم يشير إلى قيمة. نستخدم علامة `=` للإسناد (Assignment)، وهي **لا تعني "يساوي" الرياضية**، بل تعني "اجعل هذا الاسم يشير إلى هذه القيمة".

قواعد التسمية: حروف وأرقام وشرطة سفلية `_` ولا يبدأ برقم، وحساس لحالة الأحرف. والأسلوب المعتمد (PEP 8) هو `snake_case`.

</div>

```python
price: float = 120.5
quantity: int = 3
total = price * quantity
print(f"{total=}")

# يمكن إعادة إسناد المتغير بقيمة جديدة
quantity = quantity + 2
quantity += 1   # اختصار للسطر السابق
print(f"{quantity=}")
```

<div dir="rtl">

## 3. المُعامِلات (Operators)

| الفئة | المُعامِلات | مثال |
| :-- | :-- | :-- |
| حسابية | `+ - * / // % **` | `7 // 2` ← `3` |
| مقارنة | `== != < > <= >=` | `3 == 3.0` ← `True` |
| منطقية | `and or not` | `True and False` ← `False` |

انتبه: `/` تعطي دائماً `float`، و`//` قسمة صحيحة، و`%` باقي القسمة، و`**` الأُس.

</div>

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(3 == 3.0, 3 != 4, "a" < "b")
age = 20
print(age >= 18 and age < 65)

# فخّ شهير في الأعداد العشرية: ابحث عن السبب بنفسك!
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
```

<div dir="rtl">

## 4. التحويل بين الأنواع (Type Conversion)

الدالة `input()` تُرجع دائماً نصاً. لتحويله استخدم `int()` أو `float()` أو `str()`. إذا فشل التحويل ستظهر `ValueError`، وتذكّر: اقرأ السطر الأخير من رسالة الخطأ.

</div>

```python
text_number = "25"
print(int(text_number) + 5)      # 30
print(text_number + "5")         # 255 (دمج نصوص)
print(float("3.5") * 2)
print(int(7.9))                  # 7 (الاقتطاع لا التقريب)

try:
    int("abc")
except ValueError as error:
    print(f"خطأ متوقع: {error}")
```
