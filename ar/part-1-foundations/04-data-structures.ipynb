## 1. القوائم والصفوف (Lists & Tuples)

القائمة `[]` قابلة للتعديل، والصف `()` غير قابل للتعديل.

</div>

```python
langs = ["python", "go", "rust"]
langs.append("c")
langs[0] = "Python"
print(langs, langs[-1], langs[1:3], len(langs))

point = (3, 4)
x, y = point               # unpacking
print(x, y)

try:
    point[0] = 9
except TypeError as error:
    print(f"tuples are immutable: {error}")
```

<div dir="rtl">

## 2. القواميس والمجموعات (Dicts & Sets)

القاموس يربط مفتاحاً بقيمة، والمجموعة تحتفظ بالعناصر الفريدة فقط.

</div>

```python
person = {"name": "Ziad", "age": 30}
person["city"] = "Riyadh"
print(person.get("email", "n/a"))
for key, value in person.items():
    print(key, "->", value)

a, b = {1, 2, 3}, {3, 4}
print(a | b, a & b, a - b)
print(set([1, 1, 2, 2, 3]))
```

<div dir="rtl">

## 3. القوائم المختصرة (Comprehensions)

صياغة أنيقة لبناء قائمة أو قاموس في سطر واحد.

</div>

```python
squares = [n * n for n in range(1, 8) if n % 2]
lengths = {w: len(w) for w in ["apple", "fig", "banana"]}
print(squares)
print(lengths)

# Modern typing: built-in generics
scores: dict[str, list[int]] = {"ali": [90, 85], "sara": [78]}
print({k: sum(v) / len(v) for k, v in scores.items()})
```

<div dir="rtl">

### ⚠️ أخطاء شائعة

1. **`IndexError`:** الفهرس يبدأ من `0`، وآخر عنصر فهرسه `len(x) - 1` أو `-1`.
2. **`KeyError`:** استخدم `dict.get(key, default)` عندما لا تضمن وجود المفتاح.
3. **النسخ بالإسناد:** `b = a` لا تنسخ القائمة، بل تجعل `b` اسماً آخر لها. استخدم `a.copy()`.
4. **تعديل قائمة أثناء المرور عليها**، فأنشئ قائمة جديدة بدلاً من ذلك.

</div>

```python
# IndexError
nums = [1, 2, 3]
try:
    nums[3]
except IndexError as error:
    print("IndexError:", error)

# KeyError
user = {"name": "Ali"}
try:
    user["age"]
except KeyError as error:
    print("KeyError:", error)
print(user.get("age", "unknown"))

# الإسناد لا ينسخ: الاسمان يشيران إلى نفس القائمة
a = [1, 2]
b = a
b.append(3)
print(a)
# النسخ الحقيقي
c = a.copy()
c.append(4)
print(a, c)
```

<div dir="rtl">

### 🧠 نصيحة التعلّم الذاتي: جرّب لتتأكد

عندما تشك في سلوك ما (هل `b = a` تنسخ؟)، **لا تخمّن ولا تسأل أولاً؛ اكتب تجربة من ثلاثة أسطر**. البرمجة تمنحك مختبراً مجانياً بلا حدود.

</div>

<div dir="rtl">

### 📌 ملخص الفصل

- القائمة قابلة للتعديل والصف لا.
- القاموس يربط مفتاحاً بقيمة، والمجموعة تحتفظ بالفريد فقط.
- القوائم المختصرة تبني مجموعات في سطر واحد.
- الإسناد يربط اسماً بكائن ولا ينسخه.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| قابل للتعديل (Mutable) | يمكن تغيير محتواه بعد الإنشاء |
| مفتاح (Key) | ما يُستخدم للبحث في القاموس |
| تفكيك (Unpacking) | `x, y = point` |
| نسخة سطحية (Shallow copy) | نسخة لا تنسخ الكائنات الداخلية |
