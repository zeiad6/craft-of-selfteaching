## 1. الجمل الشرطية (Conditional Statements)

بايثون تحدّد الكتل **بالإزاحة (Indentation)** — أربع مسافات — لا بالأقواس. القراءة الدقيقة هنا حاسمة!

</div>

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"{score=} -> {grade=}")
```

<div dir="rtl">

## 2. حلقات التكرار (Loops)

- `for` للمرور على كائن قابل للتكرار (Iterable) مثل `range()` أو نص أو قائمة.
- `while` للتكرار ما دام الشرط صحيحاً — احذر الحلقات اللانهائية!

</div>

```python
for i in range(1, 6):
    if i == 2:
        continue          # skip 2
    if i == 5:
        break             # stop before 5
    print(i)

n = 27
steps = 0
while n != 1:             # Collatz sequence
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print(f"{steps=}")
```

## 3. مطابقة الأنماط الهيكلية (`match / case`) — Python 3.10+

بديل أوضح من سلسلة `elif` طويلة، ويستطيع تفكيك البيانات أيضاً.

</div>

```python
def describe(command: str | tuple) -> str:
    match command:
        case "start" | "run":
            return "starting"
        case ("move", x, y):
            return f"moving to ({x}, {y})"
        case ("say", *words):
            return " ".join(words)
        case _:
            return "unknown command"


print(describe("run"))
print(describe(("move", 3, 4)))
print(describe(("say", "hello", "world")))
print(describe(42))
```

### ⚠️ أخطاء شائعة

1. **`=` بدل `==`:** الأولى للإسناد والثانية للمقارنة.
2. **نسيان الإزاحة** داخل الكتلة بعد `:`.
3. **`x == 1 or 2`:** الجزء `2` قيمة صحيحة دائماً، فالشرط يتحقق دائماً. اكتب `x in (1, 2)`.
4. **حلقة `while` لا نهائية** لأنك نسيت تحديث المتغير داخلها.

</div>

```python
# الخطأ 1: = بدل ==
try:
    compile("if x = 3:\n    pass", "<demo>", "exec")
except SyntaxError as error:
    print("SyntaxError:", error.msg)

# الخطأ 2: غياب الإزاحة
try:
    compile("if True:\nprint('hi')", "<demo>", "exec")
except SyntaxError as error:
    print(type(error).__name__ + ":", error.msg)

# الخطأ 3: or مع قيمة مجردة تُعدّ دائماً صحيحة
x = 5
print(x == 1 or 2)        # 2 -> truthy, not False!
# الطريقة الصحيحة
print(x in (1, 2))        # False
```

### 📌 ملخص الفصل

- `if / elif / else` تختار مساراً واحداً من عدة مسارات، والكتل تُحدَّد بالإزاحة.
- `for` للمرور على قابل للتكرار، و`while` ما دام الشرط صحيحاً.
- `break` توقف الحلقة و`continue` تتخطى الدورة الحالية.
- `match / case` تفكّك البيانات وتطابق الأنماط (Python 3.10+).
