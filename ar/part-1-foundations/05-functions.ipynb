## 1. تعريف الدالة

الدالة كتلة قابلة لإعادة الاستخدام. نكتب **تلميحات الأنواع** و**Docstring** لتوثيقها.

</div>

```python
def area(width: float, height: float = 1.0) -> float:
    """Return the area of a rectangle."""
    return width * height


print(area(3, 4))
print(area(5))                    # default height
print(area(height=2, width=6))    # keyword arguments
help(area)
```

<div dir="rtl">

## 2. معاملات مرنة والنطاق

`*args` تجمع المعاملات الموضعية في صف، و`**kwargs` تجمع المسمّاة في قاموس. المتغير داخل الدالة محلي (Local) ولا يؤثر خارجها.

</div>

```python
def summarize(*numbers: int, **options: str) -> str:
    label = options.get("label", "total")
    return f"{label}: {sum(numbers)}"


print(summarize(1, 2, 3))
print(summarize(10, 20, label="sum"))

x = "global"


def shadow() -> None:
    x = "local"
    print(x)


shadow()
print(x)
```

## 3. دوال `lambda` والمُولِّدات

المُولِّد (Generator) ينتج القيم عند الطلب بدل بناء قائمة كاملة في الذاكرة.

</div>

```python
people = [("ali", 31), ("sara", 25), ("omar", 28)]
print(sorted(people, key=lambda p: p[1]))


def countdown(n: int):
    while n > 0:
        yield n
        n -= 1


print(list(countdown(5)))
print(sum(i * i for i in range(1_000_000)))   # generator expression
```

<div dir="rtl">

### ⚠️ أخطاء شائعة

1. **قيمة افتراضية قابلة للتعديل:** `def f(x, bucket=[])` تُنشأ القائمة **مرة واحدة** وتُشارَك بين الاستدعاءات. استخدم `None`.
2. **نسيان `return`:** الدالة تُرجع `None` ضمنياً.
3. **تعديل متغير عام داخل الدالة** دون `global` فيحدث `UnboundLocalError`. والأفضل تمرير القيمة وإرجاع النتيجة.
4. **دوال تفعل أكثر من شيء**؛ اجعل لكل دالة مسؤولية واحدة.

</div>

```python
# الخطأ 1: القيمة الافتراضية المشتركة
def add_item(item, bucket=[]):
    bucket.append(item)
    return bucket


print(add_item(1), add_item(2))   # [1, 2] [1, 2]

# الحل
def add_item_fixed(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket


print(add_item_fixed(1), add_item_fixed(2))   # [1] [2]

# الخطأ 2: لا يوجد return
def forgot_return(a, b):
    a + b


print(forgot_return(1, 2))   # None

# الخطأ 3: UnboundLocalError
counter = 0


def increment():
    counter += 1


try:
    increment()
except UnboundLocalError as error:
    print("UnboundLocalError:", error)
```

### 📌 ملخص الفصل

- الدالة تقسّم المشكلة إلى أجزاء صغيرة قابلة للاختبار.
- `*args` و`**kwargs` تجمعان المعاملات الموضعية والمسماة.
- المتغيرات داخل الدالة محلية.
- المُولِّد (`yield`) ينتج القيم عند الطلب.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| معامل (Parameter) | اسم في تعريف الدالة |
| مُعطى (Argument) | القيمة الممرّرة عند الاستدعاء |
| النطاق (Scope) | المجال الذي يكون فيه الاسم مرئياً |
| Docstring | نص التوثيق في أول الدالة |
