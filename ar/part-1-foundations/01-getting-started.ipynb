## 1. لماذا تفشل القراءة السريعة (Skimming) في تعلّم البرمجة؟

في حياتنا اليومية، اعتدنا على "القراءة السريعة" أو القفز بالعين فوق الأسطر عند تصفح المقالات ومنصات التواصل الاجتماعي. إذا غاب عنك حرف أو علامة ترقيم في مقال صحفي، فإن عقلك البشري يملأ الفراغ تلقائياً ويفهم المعنى العام.

**أما الحاسوب (Computer) فكائن حرفيّ إلى أقصى درجة!**
- مفسّر بايثون (Python Interpreter) لا يخمّن نيّتك.
- الفرق بين قوس مفتوح `(` وقوس معقوف `[`، أو بين حرف كبير `P` وحرف صغير `p`، هو الفرق بين برنامج يعمل بنجاح وبرنامج يتوقف فوراً.

لذلك، فإن **المهارة الأولى** في حرفة التعلّم الذاتي هي: **تدريب عينك وعقلك على القراءة الهادئة، كلمةً بكلمة، ورمزاً برمز.**

## 2. سطرك البرمجي الأول: الدالة `print()` والتعليقات (`Comments`)

في بايثون، عندما نريد من البرنامج أن يعرض نصاً أو نتيجة على الشاشة، نستخدم **دالة الطباعة `print()`**.

لاحظ أيضاً الرمز `#`؛ أي نص يأتي بعد `#` في نفس السطر يُسمّى **تعليقاً برمجياً (Comment)**. المفسّر يتجاهل التعليقات تماماً، لكننا نكتبها لأنفسنا وللمبرمجين الآخرين لشرح الهدف من الكود.

</div>

````python
# هذا السطر تعليق برمجي (Comment) لا ينفّذه الحاسوب
print("مرحباً بعالم التعلّم الذاتي!")
print("Hello, World!")

# يمكن لدالة print() طباعة عدة عناصر مفصولة بفاصلة، أو إجراء عمليات حسابية مباشرة:
print("نتيجة جمع 25 و 17 هي:", 25 + 17)
````

## 3. التحديث العصري (Python 3.12+): النصوص المنسّقة (`f-strings`) وتلميحات الأنواع (`Type Hints`)

في الإصدارات القديمة من بايثون، كان دمج المتغيرات داخل النصوص يتطلب طرقاً مطوّلة مثل `%` أو `.format()`. أما في بايثون الحديثة، فنستخدم **النصوص المنسّقة (`f-strings` — Formatted String Literals)** بوضع حرف `f` قبل علامة التنصيص، ووضع المتغيرات أو العمليات داخل أقواس معقوفة `{}`.

كما نستخدم في **Python 3.12+** ما يُعرف بـ **تلميحات الأنواع (Type Hints)** مثل `name: str` أو `year: int` لتوضيح نوع البيانات التي يحملها المتغير، مما يجعل الكود أوضح وأسهل قراءة:

</div>

````python
# تعريف متغيرات مع تلميحات الأنواع (Type Hints)
book_title: str = "فن التعلّم الذاتي"
target_python: float = 3.12
chapters_count: int = 12

# استخدام f-strings لدمج المتغيرات والعمليات بسلاسة
print(f"📖 الكتاب: {book_title} (محدّث لإصدار Python {target_python}+)")
print(f"🗂️ عدد الفصول: {chapters_count} فصلاً، أي بمعدل {chapters_count * 3} ساعة تدريبية تقريباً.")

# ✨ ميزة رائعة في بايثون الحديثة: طباعة اسم المتغير وقيمته معاً لأغراض الفحص باستخدام علامة (=) داخل f-string:
print(f"فحص سريع للمتغيرات: {chapters_count=}, {target_python=}")
````

## 4. كيف تقرأ رسائل الخطأ (`Traceback`) في Python 3.12+ كالمحترفين؟

عندما تخطئ في كتابة الكود (وهو أمر يحدث يومياً حتى لأمهر مهندسي البرمجيات في العالم)، يُخرج لك مفسّر بايثون تقريراً يُسمّى **تتبّع مسار الخطأ (`Traceback`)**.

الميزة الرائعة في **Python 3.11 و 3.12+** هي أن مفسّر بايثون أصبح ذكياً جداً: فهو لا يكتفي بذكر رقم السطر، بل يضع خطاً (`^^^^`) تحت الموضع الدقيق للخطأ، ويقترح عليك الحل أحياناً!

جرّب قراءة المثال التوضيحي التالي لرسالة خطأ شائعة عند كتابة `pirnt` بدلاً من `print`:

```text
NameError: name 'pirnt' is not defined. Did you mean: 'print'?
```

لاحظ كيف سألك المفسّر بلطف: **`Did you mean: 'print'?`** (هل كنت تقصد `print`؟).
لذلك، **القاعدة الذهبية عند ظهور أي خطأ هي:** اقرأ السطر الأخير من رسالة الخطأ بعناية، فهو يحتوي على نوع الخطأ ووصفه المباشر!
