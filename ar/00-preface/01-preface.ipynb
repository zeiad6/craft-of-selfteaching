## 1. لماذا نسمّي التعلّم الذاتي "حِرفة" (Craft)؟

يظنّ كثير من الناس أن القدرة على التعلّم الذاتي موهبة فطرية يولد بها البعض ويُحرم منها آخرون. الحقيقة أبسط وأجمل من ذلك بكثير: **التعلّم الذاتي حِرفة وصنعة (Craft)**، ومثل أي حِرفة أخرى — كالنجارة، أو السباحة، أو العزف، أو الطبخ — لا يمكنك إتقانها بمجرّد الاستماع إلى المحاضرات، بل بالتدريب العملي والممارسة المباشرة.

لكن لماذا يخاف معظم الناس من البدء في تعلّم شيء جديد بمفردهم، وخاصة **البرمجة (Programming)**؟

هناك ثلاثة أوهام رئيسية تقف حجر عثرة أمام المتعلّم الذاتي:

1. **وهم "الاستعداد الكامل":** الاعتقاد بأنك تحتاج إلى إتقان الرياضيات المتقدمة أو علوم الحاسب النظرية قبل أن تكتب سطرك البرمجي الأول.
2. **وهم "التعلّم الخطّي" (Linear Learning Myth):** الظنّ بأن عليك فهم كل كلمة وكل تفصيل في الفصل الأول بنسبة 100% قبل أن يُسمح لك بالانتقال إلى الفصل الثاني.
3. **الخوف من الخطأ (Fear of Errors):** التعامل مع رسائل الخطأ البرمجية (Error Messages) وكأنها عقاب أو دليل على الفشل، بينما هي في الحقيقة أصدق معلّم يوجّهك نحو الصواب.

## 2. القاعدة الذهبية: التعلّم الحقيقي يبدأ بـ "القفز إلى الأمام"

عندما تتعلّم مهارة مركّبة مثل البرمجة، فإن المفاهيم مترابطة كشبكة وليست مرصوصة كخيط مستقيم. في البداية، ستواجه مصطلحات أو رموزاً قد تبدو غامضة.

**نصيحتي الجوهرية لك في هذا الكتاب:**
- إذا قرأت فقرة ولم تفهمها بالكامل، **لا تتوقف عندها طويلاً ولا تُحبَط!**
- شغّل الكود، لاحظ النتيجة، وتقدّم إلى الفصل التالي.
- عندما تستخدم المفهوم مرّتين وثلاثاً في سياقات مختلفة، سيتبدّد الغموض تلقائياً. نحن نتعلّم **بالاستخدام والتكرار الواعي (Deliberate Practice)**، لا بالحفظ المسبق.

دعنا نثبت ذلك الآن! حتى قبل أن نبدأ الفصل الأول، يمكنك تشغيل أول كود بايثون لك في الخلية التالية بالضغط على `Shift + Enter`:

</div>

```python
# شغّل هذه الخلية بالضغط على Shift + Enter
import sys

learner_name: str = "متعلّم ذاتي مبدع"
python_version: str = f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}"

print(f"مرحباً بك يا {learner_name}!")
print(f"أنت الآن تشغّل بايثون الإصدار: Python {python_version} بنجاح 🚀")
```

## 3. المراحل الثلاث لاكتساب أي مهارة تقنية

أثناء رحلتك في هذا الكتاب، ستمرّ بثلاث مراحل طبيعية:

1. **مرحلة القراءة والفهم الدقيق (Careful Reading):** تعويد عقلك على قراءة النصوص التقنية والأكواد سطراً بسطر دون قفز عشوائي.
2. **مرحلة التقليد والتعديل (Imitation & Modification):** أخذ كود يعمل، ثم تعديل أجزائه لرؤية كيف تتغير المخرجات (ماذا لو غيّرنا هذا الرقم؟ ماذا لو حذفنا هذا القوس؟).
3. **مرحلة الإنتاج المستقل (Independent Creation):** استخدام اللبنات الأساسية التي تعلّمتها لبناء أداة صغيرة تحلّ مشكلة حقيقية في حياتك اليومية أو عملك.

---

### 🧪 تجربة تفاعلية سريعة: قانون التراكم المعرفي (Compound Growth)
يشتهر المؤلف بمثال رياضي بسيط لكنه عميق: ماذا يحدث لو تحسّنت مهارتك بنسبة **1% فقط كل يوم** لمدة عام كامل مقارنة بالثبات دون تعلّم؟
لنترك بايثون يحسب لنا النتيجة:

</div>

```python
# حساب أثر التحسّن اليومي بنسبة 1% خلال 365 يوماً
daily_improvement: float = 0.01
days: int = 365

compound_growth: float = (1 + daily_improvement) ** days

print(f"التحسّن بنسبة 1% يومياً لمدة عام يضاعف قدراتك بمقدار: {compound_growth:.2f} ضعفاً!")
```

## 🎯 التحدي التمهيدي (Warm-up Challenge)

قبل الانتقال إلى الفصل الأول، المدرّب الحقيقي يبدأ بتحريك أصابعه فوراً!

**المطلوب:**
في الخلية البرمجية أدناه:
1. غيّر قيمة المتغير `my_goal` لتكتب هدفك الشخصي من تعلّم البرمجة.
2. غيّر عدد الساعات اليومية `daily_hours` إلى الوقت الذي تخطط لتخصيصه يومياً (مثلاً `1.5` ساعة).
3. شغّل الخلية لترى إجمالي الساعات التي ستستثمرها في نفسك خلال 6 أشهر (180 يوماً).

</div>

```python
# ✍️ عدّل القيم التالية ثم شغّل الخلية (Shift + Enter):
my_goal: str = "بناء أدوات ذكية وأتمتة المهام اليومية"
daily_hours: float = 1.5
days_in_six_months: int = 180

total_hours: float = daily_hours * days_in_six_months

print(f"🎯 هدفي: {my_goal}")
print(f"⏱️ باستثمار {daily_hours} ساعة يومياً، سأمتلك خبرة عملية قدرها {total_hours:.0f} ساعة خلال 6 أشهر فقط!")
```
