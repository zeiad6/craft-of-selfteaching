## 1. المصدر الأول: بايثون نفسها

قبل البحث في الإنترنت اسأل المفسّر:

- `dir(obj)` ← ما الذي يمكنني فعله بهذا الكائن؟
- `help(obj)` ← كيف أستخدمه؟
- `type(obj)` ← ما نوعه؟

</div>

```python
import inspect

print([m for m in dir(str) if not m.startswith("_")][:12])
print(inspect.signature(str.split))
print(inspect.signature(sorted))
print(str.partition.__doc__)
```

<div dir="rtl">

## 2. تشريح توقيع الدالة

انظر إلى `sorted(iterable, /, *, key=None, reverse=False)`:

| الجزء | المعنى |
| :-- | :-- |
| `iterable` | معامل مطلوب |
| `/` | ما قبله **موضعي فقط** |
| `*` | ما بعده **بالاسم فقط** (Keyword-only) |
| `key=None` | معامل اختياري له قيمة افتراضية |

لذلك `sorted(data, reverse=True)` صحيحة، أما `sorted(data, True)` فتُسبب `TypeError`. هذه معلومة تعلّمتها من التوثيق لا من الحفظ.

</div>

```python
data = [3, 1, 2]
print(sorted(data, reverse=True))
try:
    sorted(data, True)
except TypeError as error:
    print("TypeError:", error)
```

## 3. التنقل في docs.python.org

1. **Tutorial**: للتعلّم المتسلسل.
2. **Library Reference**: لكل وحدة ودالة. هنا تقضي أغلب وقتك.
3. **Language Reference**: للقواعد الدقيقة للغة.
4. **What's New in Python 3.x**: لمعرفة المزايا الجديدة (مثل `itertools.batched` في 3.12).

**وصفة قراءة سريعة:** اقرأ أول فقرة، ثم الأمثلة، ثم المعاملات، وأخيراً ملاحظة `Changed in version` لتعرف إن كانت الميزة متوفرة عندك.

</div>

<div dir="rtl">

### ⚠️ أخطاء شائعة

1. **الاعتماد على إجابات منسوخة قديمة:** محركات البحث قد تعطيك كوداً عتيقاً لبايثون 2 أو إصدارات سابقة؛ التوثيق الرسمي أحدث وأدق.
2. **تجاهل علامات التوقيع `/` و `*`:** الشرطة المائلة تعني معاملات موضعية فقط، والنجمة تعني معاملات مسماة فقط.
3. **عدم الانتباه لملاحظات الإصدار:** تجاهل قسم `Changed in version` قد يقودك لاستخدام ميزة غير مدعومة في إصدارك الحالي.
4. **القراءة الخطية من أول الصفحة لآخرها:** لا تقرأ التوثيق كرواية؛ اقرأ أول فقرة، ثم انتقل لجدول المعاملات والأمثلة التطبيقية مباشرة.

</div>

```python
import inspect

# 1. فحص توقيع الدالة بالبرمجة
sig = inspect.signature(sorted)
print("sorted() signature:", sig)

# 2. قراءة التوثيق المضمن لدالة
# Testing positional-only vs keyword-only rules
# sorted(iterable, /, *, key=None, reverse=False)
numbers = [4, 1, 8, 3]
print("Positional iterable + keyword reverse:", sorted(numbers, reverse=True))

# 3. تجربة المعاملات المسماة والموضعية
doc_snippet = str.startswith.__doc__.splitlines()[0] if str.startswith.__doc__ else ""
print("str.startswith first line:", doc_snippet)
```

المهندس المبتدئ يسأل الآخرين فوراً، بينما المهندس المحترف يفتح التوثيق الرسمي أولاً. تعويدك لنفسك على قراءة التوثيق الرسمي لـ Python سيكسبك الاستقلالية الكاملة في أي تقنية جديدة.

### 📌 ملخص الفصل

- دالة `help()` توفر توثيق أي دالة أو فئة داخل نافذة المحرر مباشرة.
- مكتبة `inspect` تكشف توقيعات الدوال وأنواع معاملاتها برمجياً.
- الشرطة المائلة `/` في توقيع الدالة تعني أن كل ما قبلها يجب تمريره بالموضع فقط.
- علامة النجمة `*` تعني أن كل المعاملات التي تليها يجب تمريرها بذكر اسمها صراحة.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| توقيع الدالة (Signature) | شكل تعريف الدالة ومعاملاتها وقيمها الافتراضية |
| معاملات موضعية (Positional-only) | معاملات تسبق `/` لا تقبل الاسم |
| معاملات مسماة (Keyword-only) | معاملات تلي `*` تتطلب ذكر الاسم |
| Docstring | النص التوثيقي المكتوب داخل تعريف الدالة |

</div>

<div dir="rtl">

### ❓ اختبر نفسك

1. ماذا تعني علامة / داخل توقيع الدالة في بايثون؟
2. ماذا تعني علامة * بمفردها داخل توقيع الدالة؟
3. كيف تستعلم عن توثيق دالة برمجياً داخل بايثون؟
4. ما هو الموقع الرسمي لتوثيق لغة بايثون؟

<details>
<summary>الإجابات</summary>

1. تعني أن المعاملات التي تسبقها هي معاملات موضعية فقط (Positional-only) ولا يمكن تمريرها بذكر اسمها.
2. تعني أن المعاملات التي تليها هي معاملات مسماة فقط (Keyword-only) ويجب ذكر اسمها عند التمرير.
3. باستخدام الدالة المدمجة `help(function_name)` أو خاصية `function_name.__doc__`.
4. الموقع هو `docs.python.org`.
