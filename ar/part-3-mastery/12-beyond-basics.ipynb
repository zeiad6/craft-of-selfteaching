## 1. المشروع: مدير مهام بسطر الأوامر

**منهجية بناء أي مشروع:** (1) حدّد أصغر نسخة مفيدة، (2) فكّكها لدوال، (3) اختبر كل جزء، (4) وسّع تدريجياً.

سنخزّن المهام في ملف JSON ونتحكم بها عبر أوامر `add` و`list` و`done`.

</div>

```python
import argparse
import json
import tempfile
from dataclasses import asdict, dataclass
from pathlib import Path


@dataclass
class Task:
    id: int
    title: str
    done: bool = False


def load(path: Path) -> list[Task]:
    if not path.exists():
        return []
    return [Task(**item) for item in json.loads(path.read_text(encoding="utf-8"))]


def save(path: Path, tasks: list[Task]) -> None:
    path.write_text(json.dumps([asdict(t) for t in tasks], ensure_ascii=False, indent=2), encoding="utf-8")


def build_parser() -> argparse.ArgumentParser:
    parser = argparse.ArgumentParser(prog="todo")
    sub = parser.add_subparsers(dest="command", required=True)
    sub.add_parser("list")
    add = sub.add_parser("add")
    add.add_argument("title")
    done = sub.add_parser("done")
    done.add_argument("id", type=int)
    return parser


def run(argv: list[str], path: Path) -> None:
    args = build_parser().parse_args(argv)
    tasks = load(path)
    match args.command:
        case "add":
            tasks.append(Task(id=len(tasks) + 1, title=args.title))
            save(path, tasks)
            print(f"added #{len(tasks)}: {args.title}")
        case "done":
            for t in tasks:
                if t.id == args.id:
                    t.done = True
            save(path, tasks)
            print(f"completed #{args.id}")
        case "list":
            for t in tasks:
                print(f"[{'x' if t.done else ' '}] {t.id}. {t.title}")


store = Path(tempfile.mkdtemp()) / "tasks.json"
run(["add", "Read chapter 12"], store)
run(["add", "اكتب مشروعك الأول"], store)
run(["done", "1"], store)
run(["list"], store)
```

</div>

```python
import argparse

# 1. هيكل نموذج أولي بسيط
# Building a tiny CLI calculator with argparse
parser = argparse.ArgumentParser(description="A tiny CLI arithmetic tool")
parser.add_argument("x", type=float, help="First number")
parser.add_argument("y", type=float, help="Second number")
parser.add_argument("--op", choices=["add", "mul"], default="add", help="Operation")

# 2. أداة سطر أوامر مصغرة
# Simulating CLI invocation programmatically
args = parser.parse_args(["7", "6", "--op", "mul"])
result = args.x * args.y if args.op == "mul" else args.x + args.y
print(f"CLI Calculation: {args.x} {args.op} {args.y} = {result}")

# 3. التحقق من صحة المدخلات
print("Application completed successfully!")
```
