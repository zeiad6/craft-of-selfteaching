## 1. القراءة والكتابة

عبارة `with` تُغلق الملف تلقائياً حتى لو حدث خطأ. حدّد الترميز `encoding="utf-8"` دائماً، خاصةً مع العربية على Windows. نستخدم هنا مجلداً مؤقتاً كي لا نترك ملفات على جهازك.

</div>

```python
import tempfile
from pathlib import Path

workdir = Path(tempfile.mkdtemp())
note = workdir / "note.txt"

note.write_text("سطر أول\nline two\n", encoding="utf-8")

with note.open("a", encoding="utf-8") as f:
    f.write("third line\n")

for number, line in enumerate(note.read_text(encoding="utf-8").splitlines(), 1):
    print(number, line)

print(note.name, note.suffix, note.stem, note.exists(), note.stat().st_size)
```

<div dir="rtl">

## 2. JSON وCSV

</div>

```python
import csv
import json

data = {"name": "Ziad", "skills": ["python", "git"], "level": 3}
js = workdir / "profile.json"
js.write_text(json.dumps(data, ensure_ascii=False, indent=2), encoding="utf-8")
print(json.loads(js.read_text(encoding="utf-8"))["skills"])

rows = [{"name": "ali", "score": 90}, {"name": "sara", "score": 85}]
cs = workdir / "scores.csv"
with cs.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with cs.open(newline="", encoding="utf-8") as f:
    print([r for r in csv.DictReader(f)])

for p in sorted(workdir.glob("*")):
    print(p.name)
```

<div dir="rtl">

### ⚠️ أخطاء شائعة

1. **عدم استخدام `with`:** قد يبقى الملف مفتوحاً في الذاكرة ويؤدي لتلف البيانات.
2. **نسيان الترميز `encoding='utf-8'`:** يسبب أخطاء `UnicodeDecodeError` أو تشوّه الحروف العربية خاصة في Windows.
3. **استخدام مسارات نصية صلبة:** دمج المسارات يدوياً بالسلاسل النصية يفشل عبر أنظمة التشغيل المختلفة؛ استخدم `pathlib.Path`.
4. **الخلط بين `read()` و `readlines()`:** الأولى تقرأ الملف كنص واحد، والثانية كمصفوفة أسطر.

</div>

```python
from pathlib import Path
import tempfile

# 1. خطأ نسيان إغلاق الملف بدون with
tmp_dir = Path(tempfile.mkdtemp())
sample_file = tmp_dir / "demo.txt"

# 2. القراءة الآمنة بالترميز الصحيح و pathlib
# Always use with + explicit encoding
with sample_file.open("w", encoding="utf-8") as f:
    f.write("Hello, Modern Python 3.12+!\nمرحباً بالمتعلّم الذاتي!\n")

# 3. قراءة المسار كـ Path كائن ذكي
print("Path exists:", sample_file.exists())
print("File size:", sample_file.stat().st_size, "bytes")
print("Content:\n" + sample_file.read_text(encoding="utf-8"))
```

### 📌 ملخص الفصل

- عبارة `with open(...)` تضمن إغلاق الملف تلقائياً حتى لو حدث استثناء.
- مكتبة `pathlib.Path` هي المعيار الحديث للتعامل مع مسارات الملفات بدلاً من `os.path`.
- `json.dumps` و `json.loads` للتعامل مع بيانات JSON المهيكلة.
- `csv.DictReader` و `csv.DictWriter` لتسهيل قراءة وكتابة الجداول.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| مدير السياق (Context Manager) | كتلة `with` التي تتولى إدارة الموارد تلقائياً |
| pathlib | مكتبة التعامل مع المسارات ككائنات |
| الترميز (Encoding) | طريقة تحويل الحروف إلى بايتات مثل UTF-8 |
| التسلسل (Serialization) | تحويل هياكل البيانات إلى نصوص كـ JSON |
