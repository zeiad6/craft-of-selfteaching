## 1. الاستثناءات (Exceptions)

اصطد الأخطاء **المحددة** التي تتوقعها، ولا تستخدم `except:` المجردة التي تُخفي الأخطاء الحقيقية.

</div>

```python
def parse_age(text: str) -> int:
    try:
        age = int(text)
    except ValueError:
        raise ValueError(f"not a number: {text!r}") from None
    else:
        if age < 0:
            raise ValueError("age cannot be negative")
        return age
    finally:
        print("checked", repr(text))


for sample in ["30", "abc", "-5"]:
    try:
        print("age =", parse_age(sample))
    except ValueError as error:
        print("error:", error)
```

## 2. التنقيح (Debugging)

استخدم `logging` بدل `print` في المشاريع الجادة؛ فهو يدعم المستويات ويمكن إيقافه. وللتوقف التفاعلي اكتب `breakpoint()` داخل الكود.

</div>

```python
import logging

logging.basicConfig(level=logging.DEBUG, format="%(levelname)s: %(message)s", force=True)
log = logging.getLogger("demo")


def average(values: list[float]) -> float:
    log.debug("values=%s", values)
    if not values:
        log.warning("empty list, returning 0")
        return 0.0
    return sum(values) / len(values)


print(average([1, 2, 3]))
print(average([]))
```

## 3. اختبارات الوحدات

اكتب اختباراً يصف السلوك المتوقع **قبل** إصلاح الخطأ. `unittest` مضمّنة، و`pytest` أبسط صياغة (`pip install pytest`) وتكفي فيها دوال عادية و`assert`.

</div>

```python
import unittest


def is_leap(year: int) -> bool:
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


class TestLeap(unittest.TestCase):
    def test_common(self):
        self.assertTrue(is_leap(2024))
        self.assertFalse(is_leap(2023))

    def test_century_rules(self):
        self.assertFalse(is_leap(1900))
        self.assertTrue(is_leap(2000))


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestLeap)
result = unittest.TextTestRunner(verbosity=2).run(suite)
print("all passed:", result.wasSuccessful())
```

## 💻 تحديات برمجية تطبيقية

- 🟢 **مبتدئ:** اكتب دالة قسمة آمنة تُرجع `None` عند القسمة على صفر.
- 🟡 **متوسط:** اكتب اختبارات لدالة `is_palindrome` من الفصل الخامس.
- 🔴 **متقدم:** اكتب مُزخرِفاً (Decorator) `@retry(times=3)` يعيد المحاولة عند ظهور استثناء.

</div>

```python
import functools


# 1
def safe_div(a: float, b: float) -> float | None:
    try:
        return a / b
    except ZeroDivisionError:
        return None


print(safe_div(6, 3), safe_div(1, 0))


# 2
def is_palindrome(text: str) -> bool:
    cleaned = "".join(ch.lower() for ch in text if not ch.isspace())
    return cleaned == cleaned[::-1]


class TestPalindrome(unittest.TestCase):
    def test_cases(self):
        self.assertTrue(is_palindrome("Never odd or even"))
        self.assertFalse(is_palindrome("python"))


print(unittest.TextTestRunner().run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestPalindrome)).wasSuccessful())


# 3
def retry(times: int):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as exc:
                    print(f"attempt {attempt} failed: {exc}")
                    if attempt == times:
                        raise
        return wrapper
    return decorator


calls = {"n": 0}


@retry(times=3)
def flaky():
    calls["n"] += 1
    if calls["n"] < 3:
        raise RuntimeError("temporary failure")
    return "success"


print(flaky())
```
