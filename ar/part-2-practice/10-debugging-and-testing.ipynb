## 1. الاستثناءات (Exceptions)

اصطد الأخطاء **المحددة** التي تتوقعها، ولا تستخدم `except:` المجردة التي تُخفي الأخطاء الحقيقية.

</div>

```python
def parse_age(text: str) -> int:
    try:
        age = int(text)
    except ValueError:
        raise ValueError(f"not a number: {text!r}") from None
    else:
        if age < 0:
            raise ValueError("age cannot be negative")
        return age
    finally:
        print("checked", repr(text))


for sample in ["30", "abc", "-5"]:
    try:
        print("age =", parse_age(sample))
    except ValueError as error:
        print("error:", error)
```

## 2. التنقيح (Debugging)

استخدم `logging` بدل `print` في المشاريع الجادة؛ فهو يدعم المستويات ويمكن إيقافه. وللتوقف التفاعلي اكتب `breakpoint()` داخل الكود.

</div>

```python
import logging

logging.basicConfig(level=logging.DEBUG, format="%(levelname)s: %(message)s", force=True)
log = logging.getLogger("demo")


def average(values: list[float]) -> float:
    log.debug("values=%s", values)
    if not values:
        log.warning("empty list, returning 0")
        return 0.0
    return sum(values) / len(values)


print(average([1, 2, 3]))
print(average([]))
```

## 3. اختبارات الوحدات

اكتب اختباراً يصف السلوك المتوقع **قبل** إصلاح الخطأ. `unittest` مضمّنة، و`pytest` أبسط صياغة (`pip install pytest`) وتكفي فيها دوال عادية و`assert`.

</div>

```python
import unittest


def is_leap(year: int) -> bool:
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


class TestLeap(unittest.TestCase):
    def test_common(self):
        self.assertTrue(is_leap(2024))
        self.assertFalse(is_leap(2023))

    def test_century_rules(self):
        self.assertFalse(is_leap(1900))
        self.assertTrue(is_leap(2000))


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestLeap)
result = unittest.TextTestRunner(verbosity=2).run(suite)
print("all passed:", result.wasSuccessful())
```

### ⚠️ أخطاء شائعة

1. **استخدام `except:` المجردة:** يلتقط أخطاء إيقاف البرنامج (`KeyboardInterrupt`) والأخطاء المطبعية فيخفيها عنك تماماً.
2. **الاعتماد الكلي على `print` للتنقيح:** عند انتهاء المشروع تضطر لحذفها يدوياً؛ استخدم مكتبة `logging` أو `breakpoint()`.
3. **تعديل الكود قبل كتابة اختبار:** أفضل طريقة لإصلاح عطل هي كتابة اختبار يرسب أولاً، ثم إصلاح الكود حتى ينجح الاختبار.
4. **تجاهل السطر الأخير من الـ Traceback:** السطر الأخير هو الذي يحتوي دائماً على سبب المشكلة ونوعها المباشر.

</div>

```python
def safe_divide(numerator: float, denominator: float) -> float | None:
    # 1. استثناء محدد مع معالجة نظيفة
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        print("Cannot divide by zero!")
        return None
    # 2. الاستخدام الموصى به لـ try / except / else / finally
    else:
        return result
    finally:
        pass

print("10 / 2 =", safe_divide(10, 2))
print("10 / 0 =", safe_divide(10, 0))

# 3. اختبارات الوحدات الذاتية
assert safe_divide(12, 3) == 4.0
assert safe_divide(5, 0) is None
print("All assertions passed!")
```

### 🧠 نصيحة التعلّم الذاتي: الـ Traceback خريطة كنز وليست عقاباً

عندما يظهر خطأ أحمر على الشاشة، لا تنفر أو تشعر بالإحباط! اقرأ السطر الأخير أولاً لتعرف نوع الخطأ، ثم اقرأ السطور التي فوقه لتصل إلى السطر الدقيق في ملفك البرمجي الذي بدأ المشكلة.

### 📌 ملخص الفصل
