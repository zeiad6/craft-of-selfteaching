## 1. الاستيراد

بايثون تأتي مع «بطاريات مضمّنة»: مئات الوحدات الجاهزة. استكشفها بـ `dir()` و`help()`.

</div>

````python
import math
import statistics
from datetime import date, timedelta
from random import choice

print(math.sqrt(144), math.pi)
print(statistics.mean([2, 4, 9]), statistics.median([2, 4, 9]))
print(date(2026, 11, 9) + timedelta(days=30))
print(choice(["a", "b", "c"]) in "abc")
print([n for n in dir(math) if n.startswith("fa")])
````

## 2. وحدتك الخاصة

أي ملف `.py` هو وحدة. الشرط `__name__ == "__main__"` يتحقق فقط عند تشغيل الملف مباشرة لا عند استيراده. سننشئ وحدة مؤقتة ونستوردها.

</div>

````python
import importlib
import sys
import tempfile
from pathlib import Path

tmp = Path(tempfile.mkdtemp())
(tmp / "greeter.py").write_text(
    "def hello(name: str) -> str:\n"
    "    return f'Hello, {name}!'\n\n"
    "if __name__ == '__main__':\n"
    "    print('run directly')\n",
    encoding="utf-8",
)
sys.path.insert(0, str(tmp))
greeter = importlib.import_module("greeter")
print(greeter.hello("Ziad"), greeter.__name__)
````

## 3. البيئات الافتراضية والحزم الخارجية

```bash
# الطريقة الكلاسيكية
python -m venv .venv
pip install requests

# الطريقة الحديثة والأسرع (uv)
uv venv
uv pip install requests
```

سجّل المتطلبات في `requirements.txt` أو `pyproject.toml` لتتمكن من إعادة بناء البيئة.

</div>

<div dir="rtl">

### ⚠️ أخطاء شائعة

1. **تسمية ملفك بنفس اسم مكتبة قياسية:** تسمية ملفك `random.py` أو `math.py` يمنع بايثون من استيراد المكتبة الحقيقية.
2. **الاستيراد العشوائي بالرمز النجمة `from math import *`:** يلوّث نطاق المتغيرات ويصعّب معرفة مصدر الدوال.
3. **نسيان حارس التشغيل `if __name__ == '__main__':`:** يؤدي لتنفيذ أكواد الاختبار عند استيراد الملف في مكان آخر.
4. **تثبيت الحزم عالمياً دون بيئة افتراضية:** يؤدي لتعارض إصدارات المكتبات بين المشاريع المختلفة.

</div>

````python
import math
import sys

# 1. استيراد منظم ومحدد
print("Pi constant:", round(math.pi, 4))
print("Square root of 256:", math.isqrt(256))

# 2. فحص محتويات الوحدة بـ dir
trig_functions = [name for name in dir(math) if name.startswith(("sin", "cos", "tan"))]
print("Trig functions in math:", trig_functions)

# 3. استيراد مسار النظام
print("Python executable:", sys.executable)
````

### 📌 ملخص الفصل

- كل ملف بايثون `.py` يمثل وحدة برمجية (Module) يمكن استيرادها.
- المجلد الذي يحتوي على ملفات بايثون وملف `__init__.py` يمثل حزمة (Package).
- الشرط `if __name__ == '__main__':` يحدد الكود الذي يُنفّذ فقط عند تشغيل الملف مباشرة.
- البيئات الافتراضية (`venv` أو `uv`) تعزل متطلبات كل مشروع وتمنع تعارض الحزم.

### 🔤 مصطلحات الفصل

| المصطلح | المعنى |
| :-- | :-- |
| وحدة (Module) | ملف كود بايثون مستقل قابل للاستيراد |
| حزمة (Package) | مجلد ينظم عدة وحدات برمجية |
| بيئة افتراضية (Virtualenv) | مساحة معزولة لتثبيت مكتبات المشروع |
| حارس التنفيذ (Main guard) | شرط `__name__ == '__main__'` |
