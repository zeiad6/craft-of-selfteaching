## 1. الفئة والكائن

`__init__` هي الدالة التي تُنفَّذ عند إنشاء الكائن، و`self` يشير إلى الكائن نفسه.

</div>

```python
class Account:
    def __init__(self, owner: str, balance: float = 0.0) -> None:
        self.owner = owner
        self._balance = balance          # leading underscore: "internal"

    @property
    def balance(self) -> float:
        return self._balance

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("amount must be positive")
        self._balance += amount

    def __repr__(self) -> str:
        return f"Account({self.owner!r}, {self._balance})"


acc = Account("Ziad", 100)
acc.deposit(50)
print(acc, acc.balance)
```

## 2. الوراثة وتعدّد الأشكال

</div>

```python
class Shape:
    def area(self) -> float:
        raise NotImplementedError


class Square(Shape):
    def __init__(self, side: float) -> None:
        self.side = side

    def area(self) -> float:
        return self.side ** 2


class Circle(Shape):
    def __init__(self, r: float) -> None:
        self.r = r

    def area(self) -> float:
        return 3.14159 * self.r ** 2


for shape in (Square(2), Circle(1)):       # same call, different behavior
    print(type(shape).__name__, round(shape.area(), 2))
```

## 3. فئات البيانات `@dataclass`

تولّد `__init__` و`__repr__` و`__eq__` تلقائياً.

</div>

```python
from dataclasses import dataclass, field


@dataclass(order=True)
class Book:
    title: str
    pages: int
    tags: list[str] = field(default_factory=list)


a, b = Book("A", 200), Book("B", 150, ["python"])
print(a, a == Book("A", 200), a > b)
```

### ⚠️ أخطاء شائعة

1. **نسيان `self` في الدوال التابعة:** الدالة التابعة لكائن يجب أن تستقبل `self` كمعامل أول دائماً.
2. **الخلط بين متغيرات الفئة ومتغيرات النسخة:** تعديل متغير فئة مشترك قد يغير القيمة لجميع الكائنات دون قصد.
3. **نسيان استدعاء `super().__init__()`:** عند وراثة فئة أخرى، عدم استدعاء منشئ الفئة الأب يؤدي لفقدان تهيئة الخصائص.
4. **كتابة فئات ضخمة للبيانات فقط:** استخدام الكود التقليدي المطول بدلاً من `@dataclass` العصرية.

</div>

```python
class Greeter:
    def __init__(self, greeting: str = "Hello"):
        self.greeting = greeting

    # 2. الاستخدام السليم لـ self
    def greet(self, name: str) -> str:
        return f"{self.greeting}, {name}!"

g = Greeter("Welcome")
print(g.greet("Self-Learner"))

# 3. فئات البيانات الحديثة @dataclass
from dataclasses import dataclass

@dataclass
class Item:
    name: str
    price: float
    quantity: int = 1

    @property
    def total(self) -> float:
        return self.price * self.quantity

item = Item("Notebook", 15.5, 3)
print(item, f"Total: {item.total:.2f}")
```

### 📌 ملخص الفصل

- الفئة (Class) قالب مجرد، والكائن (Instance) عنصر حقيقي مبني بناءً على هذا القالب.
