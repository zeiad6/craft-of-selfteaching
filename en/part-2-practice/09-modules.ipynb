## 1. Importing

Python ships "batteries included": hundreds of ready modules. Explore them with `dir()` and `help()`.

````python
import math
import statistics
from datetime import date, timedelta
from random import choice

print(math.sqrt(144), math.pi)
print(statistics.mean([2, 4, 9]), statistics.median([2, 4, 9]))
print(date(2026, 11, 9) + timedelta(days=30))
print(choice(["a", "b", "c"]) in "abc")
print([n for n in dir(math) if n.startswith("fa")])
````

## 2. Your Own Module

Any `.py` file is a module. The `__name__ == "__main__"` guard runs only when the file is executed directly, not when imported. We'll create a temporary module and import it.

````python
import importlib
import sys
import tempfile
from pathlib import Path

tmp = Path(tempfile.mkdtemp())
(tmp / "greeter.py").write_text(
    "def hello(name: str) -> str:\n"
    "    return f'Hello, {name}!'\n\n"
    "if __name__ == '__main__':\n"
    "    print('run directly')\n",
    encoding="utf-8",
)
sys.path.insert(0, str(tmp))
greeter = importlib.import_module("greeter")
print(greeter.hello("Ziad"), greeter.__name__)
````

Record dependencies in `requirements.txt` or `pyproject.toml` so the environment can be rebuilt.

### ⚠️ Common Mistakes

1. **Shadowing standard library modules:** naming a script `math.py` or `random.py` prevents Python from finding the real standard library module.
2. **Wildcard imports (`from math import *`):** clutters your namespace and obscures where functions came from.
3. **Omitting the main guard:** code outside `if __name__ == '__main__':` executes immediately upon import.
4. **Installing packages globally:** leads to conflicting dependency versions between different projects; always use a virtual environment.

````python
import math
import sys

# 1. Explicit clean imports
print("Pi constant:", round(math.pi, 4))
print("Square root of 256:", math.isqrt(256))

# 2. Inspecting module contents with dir
trig_functions = [name for name in dir(math) if name.startswith(("sin", "cos", "tan"))]
print("Trig functions in math:", trig_functions)

# 3. System module path checking
print("Python executable:", sys.executable)
````

### 🧠 Self-Teaching Tip: Explore Batteries Included First

Before running `pip install` for an external package, check Python's standard library. It contains robust tools for math, networking, dates, hashing, compression, and testing built directly into the language.

### 📌 Chapter Summary

- Any `.py` file serves as a module that can be imported by other files.
- A folder containing an `__init__.py` file constitutes a Python package.
- The `if __name__ == '__main__':` guard isolates runnable script code from reusable library code.
- Virtual environments (`venv` or `uv`) isolate project dependencies cleanly.

### 🔤 Key Terms

| Term | Meaning |
| :-- | :-- |
| Module | a single `.py` file containing reusable Python code |
| Package | a directory containing multiple modules and an `__init__.py` |
| Virtual Environment | an isolated directory tree containing a Python installation |
| Main Guard | the idiom `if __name__ == '__main__':` |
