## 1. Reading & Writing

`with` closes the file automatically even if an error occurs. Always pass `encoding="utf-8"`, especially for Arabic on Windows. We use a temporary folder so nothing is left on your machine.

```python
import tempfile
from pathlib import Path

workdir = Path(tempfile.mkdtemp())
note = workdir / "note.txt"

note.write_text("سطر أول\nline two\n", encoding="utf-8")

with note.open("a", encoding="utf-8") as f:
    f.write("third line\n")

for number, line in enumerate(note.read_text(encoding="utf-8").splitlines(), 1):
    print(number, line)

print(note.name, note.suffix, note.stem, note.exists(), note.stat().st_size)
```

## 2. JSON & CSV

```python
import csv
import json

data = {"name": "Ziad", "skills": ["python", "git"], "level": 3}
js = workdir / "profile.json"
js.write_text(json.dumps(data, ensure_ascii=False, indent=2), encoding="utf-8")
print(json.loads(js.read_text(encoding="utf-8"))["skills"])

rows = [{"name": "ali", "score": 90}, {"name": "sara", "score": 85}]
cs = workdir / "scores.csv"
with cs.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with cs.open(newline="", encoding="utf-8") as f:
    print([r for r in csv.DictReader(f)])

for p in sorted(workdir.glob("*")):
    print(p.name)
```

### ⚠️ Common Mistakes

1. **Not using `with`:** files can stay locked in memory and leak system resources.
2. **Forgetting `encoding='utf-8'`:** leads to `UnicodeDecodeError` or garbled text, especially on Windows.
3. **Hardcoding path strings:** string concatenation breaks across OSes; always use `pathlib.Path`.
4. **Mixing up `read()` and `readlines()`:** `read()` loads the entire file as a single string, while `readlines()` returns a list of lines.

```python
from pathlib import Path
import tempfile

# 1. Unsafe file opening without with
tmp_dir = Path(tempfile.mkdtemp())
sample_file = tmp_dir / "demo.txt"

# 2. Safe reading with pathlib and utf-8
# Always use with + explicit encoding
with sample_file.open("w", encoding="utf-8") as f:
    f.write("Hello, Modern Python 3.12+!\nمرحباً بالمتعلّم الذاتي!\n")

# 3. Smart Path manipulation
print("Path exists:", sample_file.exists())
print("File size:", sample_file.stat().st_size, "bytes")
print("Content:\n" + sample_file.read_text(encoding="utf-8"))
```
