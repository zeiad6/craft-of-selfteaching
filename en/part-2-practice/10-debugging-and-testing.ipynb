## 1. Exceptions

Catch the **specific** errors you expect; avoid bare `except:` which hides real bugs.

```python
def parse_age(text: str) -> int:
    try:
        age = int(text)
    except ValueError:
        raise ValueError(f"not a number: {text!r}") from None
    else:
        if age < 0:
            raise ValueError("age cannot be negative")
        return age
    finally:
        print("checked", repr(text))


for sample in ["30", "abc", "-5"]:
    try:
        print("age =", parse_age(sample))
    except ValueError as error:
        print("error:", error)
```

## 2. Debugging

Prefer `logging` over `print` in serious projects; it supports levels and can be switched off. For interactive stepping, put `breakpoint()` in your code.

```python
import logging

logging.basicConfig(level=logging.DEBUG, format="%(levelname)s: %(message)s", force=True)
log = logging.getLogger("demo")


def average(values: list[float]) -> float:
    log.debug("values=%s", values)
    if not values:
        log.warning("empty list, returning 0")
        return 0.0
    return sum(values) / len(values)


print(average([1, 2, 3]))
print(average([]))
```

## 3. Unit Tests

Write a test describing the expected behavior **before** fixing a bug. `unittest` is built in; `pytest` has simpler syntax (`pip install pytest`)—plain functions and `assert` are enough.

```python
import unittest


def is_leap(year: int) -> bool:
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


class TestLeap(unittest.TestCase):
    def test_common(self):
        self.assertTrue(is_leap(2024))
        self.assertFalse(is_leap(2023))

    def test_century_rules(self):
        self.assertFalse(is_leap(1900))
        self.assertTrue(is_leap(2000))


suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestLeap)
result = unittest.TextTestRunner(verbosity=2).run(suite)
print("all passed:", result.wasSuccessful())
```

### ⚠️ Common Mistakes

1. **Bare `except:` blocks:** catches `KeyboardInterrupt` and typos, masking real root causes completely.
2. **Relying solely on `print` debugging:** messy to remove later; prefer Python's built-in `logging` module or `breakpoint()`.
3. **Modifying code before writing a reproducing test:** the safest debugging method is writing a failing test first, then fixing the code.
4. **Ignoring the bottom line of the Traceback:** the very last line provides the exact exception type and error description.

```python
def safe_divide(numerator: float, denominator: float) -> float | None:
    # 1. Specific exception handling
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        print("Cannot divide by zero!")
        return None
    # 2. Clean try / except / else / finally flow
    else:
        return result
    finally:
        pass

print("10 / 2 =", safe_divide(10, 2))
print("10 / 0 =", safe_divide(10, 0))

# 3. Inline assertion checks
assert safe_divide(12, 3) == 4.0
assert safe_divide(5, 0) is None
print("All assertions passed!")
```

### 🧠 Self-Teaching Tip: Tracebacks Are Maps, Not Reprimands

When a traceback appears, treat it as precise coordinates pointing directly to where your mental model diverged from reality. Start reading at the bottom, then trace upward to find your line of code.

### 📌 Chapter Summary

- `try/except` constructs prevent abnormal program termination on expected runtime faults.
- The `else` block runs only when no exception was raised; `finally` runs unconditionally.
- `breakpoint()` activates an interactive debugging session directly from your terminal.
- Automated unit testing (`unittest` / `pytest`) prevents regressions as your codebase expands.

### 🔤 Key Terms

| Term | Meaning |
| :-- | :-- |
| Exception | an error condition occurring during execution |
| Traceback | the call stack recorded when an exception is unhandled |
| Unit Test | automated verification of an isolated function or method |
| Assertion | a statement that verifies an invariant condition |
