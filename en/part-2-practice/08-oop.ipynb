## 1. Class & Object

`__init__` runs when an object is created, and `self` refers to the object itself.

```python
class Account:
    def __init__(self, owner: str, balance: float = 0.0) -> None:
        self.owner = owner
        self._balance = balance          # leading underscore: "internal"

    @property
    def balance(self) -> float:
        return self._balance

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("amount must be positive")
        self._balance += amount

    def __repr__(self) -> str:
        return f"Account({self.owner!r}, {self._balance})"


acc = Account("Ziad", 100)
acc.deposit(50)
print(acc, acc.balance)
```

## 2. Inheritance & Polymorphism

```python
class Shape:
    def area(self) -> float:
        raise NotImplementedError


class Square(Shape):
    def __init__(self, side: float) -> None:
        self.side = side

    def area(self) -> float:
        return self.side ** 2


class Circle(Shape):
    def __init__(self, r: float) -> None:
        self.r = r

    def area(self) -> float:
        return 3.14159 * self.r ** 2


for shape in (Square(2), Circle(1)):       # same call, different behavior
    print(type(shape).__name__, round(shape.area(), 2))
```

## 3. `@dataclass`

Generates `__init__`, `__repr__`, and `__eq__` automatically.

```python
from dataclasses import dataclass, field


@dataclass(order=True)
class Book:
    title: str
    pages: int
    tags: list[str] = field(default_factory=list)


a, b = Book("A", 200), Book("B", 150, ["python"])
print(a, a == Book("A", 200), a > b)
```

### ⚠️ Common Mistakes

1. **Omitting `self`:** instance methods must explicitly declare `self` as their first parameter.
2. **Confusing class variables with instance variables:** mutating a shared class attribute affects every instance unintentionally.
3. **Forgetting `super().__init__()`:** failing to call the parent initializer leaves inherited attributes uninitialized.
4. **Writing boilerplate data classes:** creating verbose `__init__` and `__repr__` methods manually when `@dataclass` exists.

```python
class Greeter:
    def __init__(self, greeting: str = "Hello"):
        self.greeting = greeting

    # 2. Proper instance method declaration
    def greet(self, name: str) -> str:
        return f"{self.greeting}, {name}!"

g = Greeter("Welcome")
print(g.greet("Self-Learner"))

# 3. Modern @dataclass in action
from dataclasses import dataclass

@dataclass
class Item:
    name: str
    price: float
    quantity: int = 1

    @property
    def total(self) -> float:
        return self.price * self.quantity

item = Item("Notebook", 15.5, 3)
print(item, f"Total: {item.total:.2f}")
```

### 🧠 Self-Teaching Tip: Don't Over-Engineer with Classes

OOP is a tool to organize complexity, not an obligation. If a task can be cleanly solved with a simple function and dictionary, prefer that over an elaborate class hierarchy.
