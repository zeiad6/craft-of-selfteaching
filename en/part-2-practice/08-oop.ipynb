## 1. Class & Object

`__init__` runs when an object is created, and `self` refers to the object itself.

```python
class Account:
    def __init__(self, owner: str, balance: float = 0.0) -> None:
        self.owner = owner
        self._balance = balance          # leading underscore: "internal"

    @property
    def balance(self) -> float:
        return self._balance

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("amount must be positive")
        self._balance += amount

    def __repr__(self) -> str:
        return f"Account({self.owner!r}, {self._balance})"


acc = Account("Ziad", 100)
acc.deposit(50)
print(acc, acc.balance)
```

## 2. Inheritance & Polymorphism

```python
class Shape:
    def area(self) -> float:
        raise NotImplementedError


class Square(Shape):
    def __init__(self, side: float) -> None:
        self.side = side

    def area(self) -> float:
        return self.side ** 2


class Circle(Shape):
    def __init__(self, r: float) -> None:
        self.r = r

    def area(self) -> float:
        return 3.14159 * self.r ** 2


for shape in (Square(2), Circle(1)):       # same call, different behavior
    print(type(shape).__name__, round(shape.area(), 2))
```

## 3. `@dataclass`

Generates `__init__`, `__repr__`, and `__eq__` automatically.

```python
from dataclasses import dataclass, field


@dataclass(order=True)
class Book:
    title: str
    pages: int
    tags: list[str] = field(default_factory=list)


a, b = Book("A", 200), Book("B", 150, ["python"])
print(a, a == Book("A", 200), a > b)
```

## 💻 Coding Challenges

- 🟢 **Warm-up:** Define a `Dog` class with `name` and a `speak()` method.
- 🟡 **Intermediate:** Add `withdraw` to `Account` that raises an error on insufficient funds.
- 🔴 **Stretch:** Design a `Stack` with `push` and `pop`, plus `__len__` and `__bool__`.

```python
class Dog:
    def __init__(self, name: str) -> None:
        self.name = name

    def speak(self) -> str:
        return f"{self.name} says woof"


print(Dog("Rex").speak())


class SafeAccount(Account):
    def withdraw(self, amount: float) -> None:
        if amount > self._balance:
            raise ValueError("insufficient funds")
        self._balance -= amount


s = SafeAccount("Sara", 10)
try:
    s.withdraw(50)
except ValueError as e:
    print("error:", e)


class Stack:
    def __init__(self) -> None:
        self._items: list = []

    def push(self, x) -> None:
        self._items.append(x)

    def pop(self):
        return self._items.pop()

    def __len__(self) -> int:
        return len(self._items)


st = Stack()
st.push(1)
st.push(2)
print(st.pop(), len(st), bool(st))
```
