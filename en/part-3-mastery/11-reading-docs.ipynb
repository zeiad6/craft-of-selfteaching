## 1. The First Source: Python Itself

Before searching the web, ask the interpreter:

- `dir(obj)` → what can I do with this object?
- `help(obj)` → how do I use it?
- `type(obj)` → what is it?

```python
import inspect

print([m for m in dir(str) if not m.startswith("_")][:12])
print(inspect.signature(str.split))
print(inspect.signature(sorted))
print(str.partition.__doc__)
```

## 2. Anatomy of a Signature

Look at `sorted(iterable, /, *, key=None, reverse=False)`:

| Part | Meaning |
| :-- | :-- |
| `iterable` | required parameter |
| `/` | everything before it is **positional-only** |
| `*` | everything after it is **keyword-only** |
| `key=None` | optional parameter with a default |

So `sorted(data, reverse=True)` is valid, while `sorted(data, True)` raises `TypeError`. You learned that from the docs, not from memorization.

```python
data = [3, 1, 2]
print(sorted(data, reverse=True))
try:
    sorted(data, True)
except TypeError as error:
    print("TypeError:", error)
```

## 3. Navigating docs.python.org

1. **Tutorial**: sequential learning.
2. **Library Reference**: every module and function. You'll spend most of your time here.
3. **Language Reference**: precise language rules.
4. **What's New in Python 3.x**: discover new features (e.g. `itertools.batched` in 3.12).

**Quick-read recipe:** first paragraph, then the examples, then the parameters, and finally the `Changed in version` notes to check the feature exists in your version.

### ⚠️ Common Mistakes

1. **Relying on outdated snippets:** web search results often surface legacy code; official docs are always up-to-date.
2. **Ignoring `/` and `*` in signatures:** `/` marks positional-only arguments and `*` marks keyword-only arguments.
3. **Skipping 'Changed in version' notes:** missing version notes causes portability issues across Python releases.
4. **Reading documentation linearly like a novel:** jump straight to the synopsis, parameters table, and runnable examples.

```python
import inspect

# 1. Inspecting function signature programmatically
sig = inspect.signature(sorted)
print("sorted() signature:", sig)

# 2. Accessing built-in docstrings
# Testing positional-only vs keyword-only rules
# sorted(iterable, /, *, key=None, reverse=False)
numbers = [4, 1, 8, 3]
print("Positional iterable + keyword reverse:", sorted(numbers, reverse=True))

# 3. Testing parameter constraints
doc_snippet = str.startswith.__doc__.splitlines()[0] if str.startswith.__doc__ else ""
print("str.startswith first line:", doc_snippet)
```

### 🧠 Self-Teaching Tip: The Official Docs Guarantee Autonomy

A beginner asks online right away; a seasoned developer opens the official documentation first. Mastering the art of reading official docs gives you lifelong independence in any language.

### 📌 Chapter Summary

- The `help()` built-in displays comprehensive documentation directly in your terminal or notebook.
- The `inspect` module provides programmatic introspection of callable signatures.
- The slash `/` indicates positional-only parameters.
- The bare asterisk `*` indicates keyword-only parameters.

### 🔤 Key Terms

| Term | Meaning |
| :-- | :-- |
| Function Signature | the formal declaration of parameters and return types |
| Positional-only (`/`) | arguments that cannot be passed by keyword name |
| Keyword-only (`*`) | arguments that must be passed with explicit keyword names |
| Docstring | the doc comment attached to a module, class, or function |
