## 1. Defining a Function

A function is a reusable block. We add **type hints** and a **docstring** to document it.

```python
def area(width: float, height: float = 1.0) -> float:
    """Return the area of a rectangle."""
    return width * height


print(area(3, 4))
print(area(5))                    # default height
print(area(height=2, width=6))    # keyword arguments
help(area)
```

## 2. Flexible Parameters & Scope

`*args` collects positional arguments into a tuple, `**kwargs` collects keyword ones into a dict. Variables inside a function are local.

```python
def summarize(*numbers: int, **options: str) -> str:
    label = options.get("label", "total")
    return f"{label}: {sum(numbers)}"


print(summarize(1, 2, 3))
print(summarize(10, 20, label="sum"))

x = "global"


def shadow() -> None:
    x = "local"
    print(x)


shadow()
print(x)
```

## 3. `lambda` and Generators

A generator yields values on demand instead of building a full list in memory.

```python
people = [("ali", 31), ("sara", 25), ("omar", 28)]
print(sorted(people, key=lambda p: p[1]))


def countdown(n: int):
    while n > 0:
        yield n
        n -= 1


print(list(countdown(5)))
print(sum(i * i for i in range(1_000_000)))   # generator expression
```

### ⚠️ Common Mistakes

1. **Mutable default argument:** in `def f(x, bucket=[])` the list is created **once** and shared across calls. Use `None`.
2. **Forgetting `return`:** the function implicitly returns `None`.
3. **Rebinding a global inside a function** without `global` causes `UnboundLocalError`. Better: pass values in and return results.
4. **Functions that do more than one thing;** give each function a single responsibility.

```python
# Mistake 1: the shared default
def add_item(item, bucket=[]):
    bucket.append(item)
    return bucket


print(add_item(1), add_item(2))   # [1, 2] [1, 2]

# The fix
def add_item_fixed(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket


print(add_item_fixed(1), add_item_fixed(2))   # [1] [2]

# Mistake 2: no return
def forgot_return(a, b):
    a + b


print(forgot_return(1, 2))   # None

# Mistake 3: UnboundLocalError
counter = 0


def increment():
    counter += 1


try:
    increment()
except UnboundLocalError as error:
    print("UnboundLocalError:", error)
```
