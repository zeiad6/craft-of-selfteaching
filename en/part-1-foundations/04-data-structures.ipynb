## 1. Lists & Tuples

A list `[]` is mutable; a tuple `()` is immutable.

```python
langs = ["python", "go", "rust"]
langs.append("c")
langs[0] = "Python"
print(langs, langs[-1], langs[1:3], len(langs))

point = (3, 4)
x, y = point               # unpacking
print(x, y)

try:
    point[0] = 9
except TypeError as error:
    print(f"tuples are immutable: {error}")
```

## 2. Dicts & Sets

A dict maps keys to values; a set keeps unique elements only.

```python
person = {"name": "Ziad", "age": 30}
person["city"] = "Riyadh"
print(person.get("email", "n/a"))
for key, value in person.items():
    print(key, "->", value)

a, b = {1, 2, 3}, {3, 4}
print(a | b, a & b, a - b)
print(set([1, 1, 2, 2, 3]))
```

## 3. Comprehensions

A concise way to build lists and dicts in one line.

```python
squares = [n * n for n in range(1, 8) if n % 2]
lengths = {w: len(w) for w in ["apple", "fig", "banana"]}
print(squares)
print(lengths)

# Modern typing: built-in generics
scores: dict[str, list[int]] = {"ali": [90, 85], "sara": [78]}
print({k: sum(v) / len(v) for k, v in scores.items()})
```

### ⚠️ Common Mistakes

1. **`IndexError`:** indexing starts at `0`; the last item is `len(x) - 1` or `-1`.
2. **`KeyError`:** use `dict.get(key, default)` when the key may be missing.
3. **Copy by assignment:** `b = a` does not copy the list; `b` becomes another name for it. Use `a.copy()`.
4. **Mutating a list while iterating over it**; build a new list instead.

```python
# IndexError
nums = [1, 2, 3]
try:
    nums[3]
except IndexError as error:
    print("IndexError:", error)

# KeyError
user = {"name": "Ali"}
try:
    user["age"]
except KeyError as error:
    print("KeyError:", error)
print(user.get("age", "unknown"))

# Assignment does not copy: both names point to one list
a = [1, 2]
b = a
b.append(3)
print(a)
# A real copy
c = a.copy()
c.append(4)
print(a, c)
```

### 🧠 Self-Teaching Tip: Experiment to Be Sure

When you doubt a behavior (does `b = a` copy?), **don't guess and don't ask first; write a three-line experiment**. Programming hands you a free, unlimited laboratory.

### 📌 Chapter Summary

- Lists are mutable; tuples are not.
- A dict maps keys to values; a set keeps unique items only.
- Comprehensions build collections in one line.
- Assignment binds a name to an object; it does not copy it.

### 🔤 Key Terms

| Term | Meaning |

| Key | what you use to look things up in a dict |
| Unpacking | `x, y = point` |
