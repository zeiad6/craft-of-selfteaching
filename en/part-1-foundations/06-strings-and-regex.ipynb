## 1. Basic String Operations

Python strings are fully Unicode, so Arabic works out of the box. Strings are **immutable**; every method returns a new string.

```python
s = "  Self-Teaching is a Craft  "
print(s.strip().lower())
print(s.split())
print("-".join(["a", "b", "c"]))
print("Craft" in s, s.find("Craft"), s.strip()[:4], s.strip()[::-1])
print("مرحباً بالعالم".upper(), len("مرحباً"))
print(f"{3.14159:.2f} | {1234567:,} | {'x':>5} | {0.256:.1%}")
```

## 2. Regular Expressions

A regex is a mini-language for describing **text patterns**. Use raw strings `r"..."` so backslashes aren't interpreted.

| Token | Meaning |
| :-- | :-- |
| `\d` | digit |
| `\w` | letter, digit or `_` |
| `+` / `*` / `?` | one+ / zero+ / optional |
| `[...]` | any one of the characters |
| `(...)` | capture group |

```python
import re

text = "Contact: ali@example.com, sara.k@mail.org | Tel: 2026-11-09"

emails = re.findall(r"[\w.]+@[\w.]+\.\w+", text)
print(emails)

m = re.search(r"(\d{4})-(\d{2})-(\d{2})", text)
if m:
    year, month, day = m.groups()
    print(f"{year=} {month=} {day=}")

print(re.sub(r"\d", "#", "Phone 0501234567"))
```

### ⚠️ Common Mistakes

1. **Trying to modify a string:** `word[0] = "P"` raises `TypeError`; strings are immutable.
2. **Greedy matching:** `.+` swallows as much as possible. Use `.+?` for the shortest match.
3. **Forgetting raw strings:** write `r"\d+"`, not `"\d+"`.
4. **Over-clever regexes:** if you can't read it, use `split` / `startswith`, or break it up with `re.VERBOSE`.

```python
import re

# Strings are immutable
word = "python"
try:
    word[0] = "P"
except TypeError as error:
    print("TypeError:", error)
print(word.capitalize())

# Greedy vs lazy
html = "<b>bold</b> and <i>italic</i>"
print(re.findall(r"<.+>", html))
print(re.findall(r"<.+?>", html))

# Raw strings keep backslashes
print(len("\n"), len(r"\n"))
```

### 🧠 Self-Teaching Tip: Build Test Examples

Before writing any regex, collect **5 strings that must match and 5 that must not**. Re-test with them every time you tweak the pattern—you'll have built a tiny test suite while learning.

### 📌 Chapter Summary

### 🔤 Key Terms

| Term | Meaning |
| :-- | :-- |
| Raw string | `r".."` does not interpret backslashes |
| Capture group | `(...)` extracts part of a match |
| Greedy | matches as much as possible |
| Unicode | a standard covering Arabic, emoji and more |
