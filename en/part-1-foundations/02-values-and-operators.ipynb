## 1. Values and Data Types

A **value** is the simplest unit of data: a number, a piece of text, true/false. Every value has a **type** that determines what you can do with it. Don't memorize types—**ask Python** with `type()`. That is self-teaching in action.

```python
print(type(42))        # int      : whole number
print(type(3.14))      # float    : decimal number
print(type("hello"))   # str      : text
print(type(True))      # bool     : boolean
print(type(None))      # NoneType : absence of a value
```

## 2. Variables

A **variable** is a name that refers to a value. The `=` sign is **assignment**, not mathematical equality: it means "make this name point to this value."

Naming rules: letters, digits, and underscores; cannot start with a digit; case-sensitive. The PEP 8 convention is `snake_case`.

```python
price: float = 120.5
quantity: int = 3
total = price * quantity
print(f"{total=}")

# Variables can be reassigned
quantity = quantity + 2
quantity += 1   # shorthand for the line above
print(f"{quantity=}")
```

## 3. Operators

| Category | Operators | Example |
| :-- | :-- | :-- |
| Arithmetic | `+ - * / // % **` | `7 // 2` → `3` |
| Comparison | `== != < > <= >=` | `3 == 3.0` → `True` |
| Logical | `and or not` | `True and False` → `False` |

Note: `/` always returns a `float`, `//` is floor division, `%` is the remainder, `**` is exponentiation.

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(3 == 3.0, 3 != 4, "a" < "b")
age = 20
print(age >= 18 and age < 65)

# A famous floating-point trap: research why it happens!
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))
```

## 4. Type Conversion

`input()` always returns a string. Convert with `int()`, `float()`, or `str()`. A failed conversion raises `ValueError`—and remember: read the last line of the traceback.

```python
text_number = "25"
print(int(text_number) + 5)      # 30
print(text_number + "5")         # 255 (string concatenation)
print(float("3.5") * 2)
print(int(7.9))                  # 7 (truncates, doesn't round)

try:
    int("abc")
except ValueError as error:
    print(f"Expected error: {error}")
```
