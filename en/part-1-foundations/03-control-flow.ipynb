## 1. Conditional Statements

Python marks blocks by **indentation**—four spaces—not braces. Careful reading matters here!

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(f"{score=} -> {grade=}")
```

## 2. Loops

- `for` walks over an iterable such as `range()`, a string, or a list.
- `while` repeats as long as a condition holds—beware infinite loops!

```python
for i in range(1, 6):
    if i == 2:
        continue          # skip 2
    if i == 5:
        break             # stop before 5
    print(i)

n = 27
steps = 0
while n != 1:             # Collatz sequence
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print(f"{steps=}")
```

## 3. Structural Pattern Matching (`match / case`) — Python 3.10+

A clearer alternative to long `elif` chains that can also destructure data.

```python
def describe(command: str | tuple) -> str:
    match command:
        case "start" | "run":
            return "starting"
        case ("move", x, y):
            return f"moving to ({x}, {y})"
        case ("say", *words):
            return " ".join(words)
        case _:
            return "unknown command"


print(describe("run"))
print(describe(("move", 3, 4)))
print(describe(("say", "hello", "world")))
print(describe(42))
```

### ⚠️ Common Mistakes

1. **`=` instead of `==`:** the first assigns, the second compares.
2. **Forgetting indentation** inside a block after `:`.
3. **`x == 1 or 2`:** the bare `2` is always truthy, so the condition always passes. Write `x in (1, 2)`.
4. **Infinite `while` loops** because the loop variable is never updated.

```python
# Mistake 1: = instead of ==
try:
    compile("if x = 3:\n    pass", "<demo>", "exec")
except SyntaxError as error:
    print("SyntaxError:", error.msg)

# Mistake 2: missing indentation
try:
    compile("if True:\nprint('hi')", "<demo>", "exec")
except SyntaxError as error:
    print(type(error).__name__ + ":", error.msg)

# Mistake 3: a bare value after 'or' is always truthy
x = 5
print(x == 1 or 2)        # 2 -> truthy, not False!
# The correct way
print(x in (1, 2))        # False
```

### 🧠 Self-Teaching Tip: Predict Before You Run

Before running any cell, write down **what you expect it to print**. If the result matches, you understood; if it doesn't, you've found a gap in your mental model—the most valuable thing you can discover.

### 📌 Chapter Summary

- `if / elif / else` picks one path among several; blocks are defined by indentation.
- `for` walks an iterable; `while` repeats while a condition holds.
- `break` stops a loop; `continue` skips the current iteration.
- `match / case` destructures data and matches patterns (Python 3.10+).

### 🔤 Key Terms

| Term | Meaning |
| :-- | :-- |
| Conditional | a block that runs when a condition holds |
| Iterable | anything whose items you can walk through |
| Indentation | the spaces that mark a block's boundaries |
| Pattern matching | `match / case` |
