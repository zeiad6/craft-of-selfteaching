## 1. Why Call Self-Teaching a "Craft"?

Many people assume that the ability to teach oneself is an innate gift reserved for a lucky few. The reality is far more empowering: **self-teaching is a craft**. Like woodworking, swimming, playing an instrument, or cooking, a craft cannot be mastered by passively listening to lectures—it is honed through hands-on practice, experimentation, and iteration.

So why do so many intelligent adults hesitate when trying to learn a new technical skill—especially **programming**—on their own?

Three common illusions stand in the way:

1. **The Illusion of "Total Readiness":** Believing you must master advanced mathematics or theoretical computer science before writing your very first line of code.
2. **The Myth of Linear Learning:** Assuming you must understand 100% of Chapter 1 before you are allowed to turn the page to Chapter 2.
3. **The Fear of Error Messages:** Treating a Python `Traceback` or error message as a personal failure, rather than what it truly is: objective, immediate feedback showing you exactly how to adjust your mental model.

## 2. The Golden Rule: Lean Forward and Keep Moving

When learning a complex system like programming, concepts are interconnected like a web rather than arranged in a single straight line. Early on, you will inevitably encounter terms or syntax that feel slightly unfamiliar.

**Our core rule throughout this book:**
- If a paragraph doesn't click 100% on your first read, **do not stall or get discouraged**.
- Run the code, observe what happens, and **keep moving forward**.
- Once you use a concept two or three times in different contexts, the fog lifts naturally. True mastery comes from **deliberate usage**, not upfront memorization.

## 3. The Three Stages of Acquiring Any Technical Craft

As you work through this book, you will naturally progress through three stages:

1. **Careful Reading:** Training your attention to read technical explanations and code line by line without skimming over crucial details.
2. **Imitation & Tinkering:** Taking working code and deliberately modifying it (*"What happens if I change this number? What if I remove this parenthesis?"*).
3. **Independent Creation:** Combining the building blocks you have practiced to build small, useful programs that solve real problems.

### 🧪 Interactive Insight: The Power of Compound Learning
A classic mental model from the book illustrates what happens when you improve your craft by just **1% every day** over a year (`365` days). Let's let Python calculate the compound effect:

```python
# Compounding a 1% daily improvement over 365 days
daily_improvement: float = 0.01
days: int = 365

compound_growth: float = (1 + daily_improvement) ** days

print(f"Improving by just 1% a day for a year multiplies your capability by {compound_growth:.2f}x!")
```
